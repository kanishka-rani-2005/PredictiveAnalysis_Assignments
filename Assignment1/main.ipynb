```python
#import samplers from imblearn
from imblearn.under_sampling import RandomUnderSampler
from imblearn.over_sampling import RandomOverSampler, SMOTE, ADASYN
from imblearn.combine import SMOTETomek
from imblearn.under_sampling import NearMiss
from imblearn.under_sampling import TomekLinks
```

```python
#prepare x and y 
# drop target column from training dataset
X = df.drop("Class", axis=1)
y = df["Class"]
```

```python
# 5 samplers we need to use
samplers = {
    "Undersampling": RandomUnderSampler(random_state=42),
    "Oversampling": RandomOverSampler(random_state=42),
    "SMOTE": SMOTE(random_state=42),
    "ADASYN": ADASYN(random_state=42),
    "SMOTE-Tomek": SMOTETomek(random_state=42),
    "NearMiss": NearMiss(),
    "TomekLinks": TomekLinks()
}
```

```python
#generate balanced datasets
balanced_data = {} #dictionary to store balanced datasets for each sampler
for name, sampler in samplers.items():
    X_res, y_res = sampler.fit_resample(X, y)
    balanced_data[name] = (X_res, y_res)
    # print(f"{name} sampling: {y_res.value_counts()}")  #for checking
```

```python
#classification probelem So we will import 5 classifiers
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
```

```python
models = {
    "LogisticReg": LogisticRegression(max_iter=1000),
    "DecisionTree": DecisionTreeClassifier(),
    "RandomForest": RandomForestClassifier(),
    "SVM": SVC(),
    "KNN": KNeighborsClassifier()
}
```

```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

# model name will be  name and sampler name will be column name
results=pd.DataFrame(index=models.keys(), columns=samplers.keys())


for sampler ,(X_res,y_res) in balanced_data.items():
    X_train, X_test, y_train, y_test = train_test_split(X_res, y_res, test_size=0.3, random_state=42)
    for model_name, model in models.items():
        model.fit(X_train, y_train) # fit the model on training dataset
        y_pred = model.predict(X_test) # predict on test dataset
        accuracy = accuracy_score(y_test, y_pred) # calculate accuracy
        results.loc[model_name, sampler] = accuracy # store accuracy in results dataframe
```
